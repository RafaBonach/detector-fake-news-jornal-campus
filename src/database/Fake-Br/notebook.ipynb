# Configurando variáveis de treinamento

```python
from sklearn.model_selection import train_test_split # Para dividir a "prova"
```

```python
# Definir quem são as perguntas (X) e quem é a resposta (y)
dataset_shuffled = dataset.sample(frac=1).reset_index(drop=True)

x = dataset_shuffled["preprocessed_news"]
y = dataset_shuffled["Classificação"]

# Dividir a "prova" em 80% para treino e 20% para teste
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print(f"Tamanho do Treino (Estudo): {x_train.shape[0]} noticias")
print(f"Tamanho do Teste (Prova): {x_test.shape[0]} noticias")

```
