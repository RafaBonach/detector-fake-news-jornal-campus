```python
# Remover todas as linhas que não tiverem título
df1 = df1.dropna(subset=['Titulo'])

df1.info()
```

```python
df2 = df2.dropna(subset=['Titulo'])

df2.info()
```

```python
# Vamos identificar o que tem na coluna "Classe"
df1['Classe'].unique()
```

```python
# Vamos transformar a coluna "Classe" em uma coluna de inteiros.
df1_int = df1.copy()
df1_int['Classe'] = df1['Classe'].astype(int)

df1_int.info()
```

```python
df1_int['Classe'].unique()
```

```python
# Salvando os dataframes
df1_int.to_csv("./FakeRecogna_tratada.csv", index=False)
df2.to_csv("./FakeRecogna_no_removal_words_tratada.csv", index=False)
```
